```python
# 🛠️ Setup: Import libraries
import pandas as pd
import time
import psutil
import os
import gc
```

```python
# Utility function to print current RAM usage
def print_memory_usage(label=""):
    process = psutil.Process(os.getpid())
    mem_MB = process.memory_info().rss / 1024 / 1024
    print(f"{label} | RAM Usage: {mem_MB:.2f} MB")
```

```python
print("\nLoading Sample Datasets (sample_collisions.csv + sample_parties.csv)")
print_memory_usage("Before loading sample")

start = time.time()
sample_collisions = pd.read_csv(
    "/content/drive/MyDrive/Datasets/Crash_Data_Analysis_Sample_Dataset/sample_collisions.csv"
)
sample_parties = pd.read_csv(
    "/content/drive/MyDrive/Datasets/Crash_Data_Analysis_Sample_Dataset/sample_parties.csv"
)
end = time.time()

print(f"Sample data loaded in {end - start:.2f} seconds")
print_memory_usage("After loading sample")

# Preview data
print(f"Sample collisions: {sample_collisions.shape}")
print(f"Sample parties: {sample_parties.shape}")
display(sample_collisions.head(2))
display(sample_parties.head(2))

# Sample ETL
sample_collisions.columns = (
    sample_collisions.columns.str.lower().str.strip().str.replace(" ", "_")
)
sample_parties.columns = (
    sample_parties.columns.str.lower().str.strip().str.replace(" ", "_")
)

sample_df = pd.merge(
    sample_collisions[["case_id", "county_location", "collision_date"]],
    sample_parties[["case_id", "party_age"]],
    on="case_id",
    how="inner",
)
sample_df["collision_date"] = pd.to_datetime(
    sample_df["collision_date"], errors="coerce"
)
sample_df["party_age"] = pd.to_numeric(sample_df["party_age"], errors="coerce").fillna(
    -1
)

print("Sample ETL complete")
display(sample_df.head(2))
```

```python
# Clear sample data from memory to free up RAM before loading large data
del sample_collisions
del sample_parties
del sample_df
gc.collect()

print("Cleared sample datasets from memory")
print_memory_usage("After clearing sample data")
```

```python
print("\n Loading Full Datasets (collisions.csv + parties.csv)")
print_memory_usage("Before loading full")

start = time.time()
full_collisions = pd.read_csv(
    "/content/drive/MyDrive/Datasets/Crash_Data_Analysis_Full_Dataset/collisions.csv"
)
full_parties = pd.read_csv(
    "/content/drive/MyDrive/Datasets/Crash_Data_Analysis_Full_Dataset/parties.csv"
)
end = time.time()

print(f"Full data loaded in {end - start:.2f} seconds")
print_memory_usage("After loading full")

# Preview
print(f"Full collisions: {full_collisions.shape}")
print(f"Full parties: {full_parties.shape}")
display(full_collisions.head(2))
display(full_parties.head(2))

# Full ETL
full_collisions.columns = (
    full_collisions.columns.str.lower().str.strip().str.replace(" ", "_")
)
full_parties.columns = (
    full_parties.columns.str.lower().str.strip().str.replace(" ", "_")
)

etl_start = time.time()

full_df = pd.merge(
    full_collisions[["case_id", "county_location", "collision_date"]],
    full_parties[["case_id", "party_age"]],
    on="case_id",
    how="inner",
)
full_df["collision_date"] = pd.to_datetime(full_df["collision_date"], errors="coerce")
full_df["party_age"] = pd.to_numeric(full_df["party_age"], errors="coerce").fillna(-1)

etl_end = time.time()
print(f"Full ETL complete in {etl_end - etl_start:.2f} seconds")
print_memory_usage("After full ETL")

display(full_df.head(2))
```
