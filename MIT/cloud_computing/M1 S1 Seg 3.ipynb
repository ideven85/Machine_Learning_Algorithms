```python
# Import libraries
import pandas as pd
```

```python
# Load the collision data
# Use the sample dataset present in the script and save it in the same directory as the notebook
collisions = pd.read_csv("sample_collisions.csv")
print("Collisions data loaded:", collisions.shape)

# Load the parties (drivers/passengers) data
parties = pd.read_csv("sample_parties.csv")
print("Parties data loaded:", parties.shape)
```

```python
# Real-world note:
# In real data projects, the transformation stage often takes 70–80% of the total effort.
# This includes cleaning column names, fixing formats, merging files, and filling missing values.
# These steps are essential before any meaningful analysis can be done.
```

```python
# Step 1: Clean column names (make them lowercase and replace spaces with underscores)
collisions.columns = collisions.columns.str.lower().str.strip().str.replace(" ", "_")
parties.columns = parties.columns.str.lower().str.strip().str.replace(" ", "_")
```

```python
print("Step 1: Cleaned column names (Collisions)")
display(collisions.head(2))
display(collisions.tail(2))
```

```python
print("Step 1: Cleaned column names (Parties)")
display(parties.head(2))
display(parties.tail(2))
```

```python
# Step 2: Select only relevant columns
collisions_subset = collisions[["case_id", "county_location"]]
parties_subset = parties[["case_id", "party_age"]]
```

```python
print("Step 2: Selected columns from collisions")
display(collisions_subset.head(2))
display(collisions_subset.tail(2))
```

```python
print("Step 2: Selected columns from parties")
display(parties_subset.head(2))
display(parties_subset.tail(2))
```

```python
# Step 3: Merge on 'case_id'
merged = pd.merge(collisions_subset, parties_subset, on="case_id", how="inner")
```

```python
print("Step 3: After merge (Collisions + Parties)")
display(merged.head(2))
display(merged.tail(2))
```

```python
# Step 4: Fix data types
merged["party_age"] = pd.to_numeric(merged["party_age"], errors="coerce")

print("Step 4: Converted party_age to numeric")
display(merged.head(2))
display(merged.tail(2))
```

```python
# Step 5: Fill missing values
merged["party_age"] = merged["party_age"].fillna(-1)

print("Step 5: Filled missing party_age values with -1")
display(merged.head(2))
display(merged.tail(2))
```

```python
# Step 6: Standardize text (title-case for county_location)
merged["county_location"] = merged["county_location"].str.title().str.strip()

print("Step 6: Standardized county_location text")
display(merged.head(2))
display(merged.tail(2))
```
