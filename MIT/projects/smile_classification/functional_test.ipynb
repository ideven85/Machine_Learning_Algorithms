```python
class SimpleNet(nn.Module):
    def __init__(self):
        super(SimpleNet, self).__init__()
        # Stateful layers: They manage their own weights and biases internally

        self.fc1 = nn.Linear(in_features=10, out_features=5)
        self.fc2 = nn.Linear(5, 2)

    def forward(self, x):
        # Stateless activation: F.relu does not have weights

        x = F.relu(self.fc1(x))

        # Stateless output activation: Softmax over the last dimension
        x = F.softmax(self.fc2(x), dim=-1)
        return x

    def __call__(self, x, *args, **kwargs):
        return self.forward(x)
```

```python
net = SimpleNet()
```
