```python
import numpy as np
import pandas as pd
```

```python
text = """Alice commented on Bob's (a new Software Engineer hire in the team) approach to solving the problem. Alice was quoted in the meeting as saying "Bob's algorithm is theoretically optimal. But, the algorithm is very hard to implement in real time. After all, he does not have an EE background and does not yet understand the way read thresholds work in a memory chip. He needs to come up to speed with what is possible in the read channel. So, set up a meeting with the hardware and chip design engineers." Alice then ended the meeting.""".split()
vocab = dict()
```

```python
for word in text:
    if word in vocab:
        vocab[word] += 1
    else:
        vocab[word] = 1
```

```python
sorted(vocab.items(), key=lambda x: x[1])
```
