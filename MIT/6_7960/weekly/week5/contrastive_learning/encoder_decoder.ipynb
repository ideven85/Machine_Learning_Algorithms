```python
from typing import Any

import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class Encoder(nn.Module):
    def __init__(self, input_dim=784, hidden_dim=128, embedding_dim=32):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, embedding_dim),
        )

    def forward(self, x):
        return self.net(x)
```

```python
class Decoder(nn.Module):
    def __init__(self, embedding_dim=32, hidden_dim=128, output_dim=784):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(embedding_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, output_dim),
        )

    def forward(self, x):
        return self.net(x)
```

```python
class AutoEncoder(nn.Module):
    def __init__(self, *args: Any, **kwargs: Any):
        super().__init__()
        self.encoder = Encoder()
        self.decoder = Decoder()

    def forward(self, x):
        learnt_representation = self.encoder(x)
        reconstruction = self.decoder(x)
        return (
            reconstruction,
            learnt_representation,
        )
```

```python
model = AutoEncoder()
optim = torch.optim.Adam(model.parameters(), lr=1e-03)
```
